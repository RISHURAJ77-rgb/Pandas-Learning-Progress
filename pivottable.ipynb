```python
import numpy as np
import pandas as pd
```

```python
data={
'date':pd.date_range('2023-01-01',periods=20),
'Product':['A','B','C','D'] *5,
'Region':['East','west','North','south','East','west','North','south','East','west','North','south',
          'East','west','North','south','East','west','North','south'],
'sales':np.random.randint(100,1000,20),
'unit':np.random.randint(10,100,20),
'Rep':['john','mary','bob','Alice','john','mary','bob','Alice','john','mary','bob','Alice',
       'john','mary','bob','Alice','john','mary','bob','Alice']
}
df=pd.DataFrame(data)
df['month']=df['date'].dt.month_name()
df['Quarter']='Q' + df['date'].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df,values='sales',index='Region',columns='Product',aggfunc='median')
```

```python
pivot2=pd.pivot_table(df, values=['sales','unit'],index='Region',columns='Product')
pivot2
```
